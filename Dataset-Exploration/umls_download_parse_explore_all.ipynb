# META Folder


```python
import os
import gzip
import re
from collections import defaultdict

source_folder = '/content/drive/MyDrive/UMLS/2025AB/META'
destination_folder = '/content/drive/MyDrive/UMLS/2025RRFs'

# Create the destination folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)

print(f"Source folder: {source_folder}")
print(f"Destination folder: {destination_folder}")

rrf_files_to_process = defaultdict(list)

# Group gzipped files by their base RRF name
for filename in os.listdir(source_folder):
    if filename.endswith('.gz'):
        # Use regex to find the base RRF name
        # This handles both MRCONSO.RRF.aa.gz and MRDEF.RRF.gz
        # Updated regex to correctly capture the base RRF name
        match = re.match(r"([A-Z_]+\.RRF)(?:\.[a-z]{2})?\.gz$", filename)
        if match:
            base_rrf_name = match.group(1)
            full_path = os.path.join(source_folder, filename)
            rrf_files_to_process[base_rrf_name].append(full_path)
        else:
            print(f"Warning: Could not parse base RRF name for {filename}. Skipping.")

print(f"\nFound {len(rrf_files_to_process)} unique RRF groups to extract.")

# Process each group of RRF files
for base_rrf_name, gzipped_parts in rrf_files_to_process.items():
    output_rrf_path = os.path.join(destination_folder, base_rrf_name)

    # Sort the parts to ensure correct order for concatenation (e.g., .aa, .ab, .ac)
    gzipped_parts.sort()

    print(f"\nExtracting and concatenating {len(gzipped_parts)} part(s) for '{base_rrf_name}' to '{output_rrf_path}'...")

    try:
        # Open the output RRF file in text write mode with UTF-8 encoding
        with open(output_rrf_path, 'wt', encoding='utf-8') as outfile:
            for part_file_path in gzipped_parts:
                print(f"  - Processing: {os.path.basename(part_file_path)}")
                # Open each gzipped part in text read mode with UTF-8 encoding
                with gzip.open(part_file_path, 'rt', encoding='utf-8') as infile:
                    outfile.write(infile.read())
        print(f"✅ Successfully extracted and concatenated '{base_rrf_name}'")
    except Exception as e:
        print(f"❌ Error processing '{base_rrf_name}': {e}")

print("\nExtraction process complete.")
```

```python
import pandas as pd
from IPython.display import display

def read_rrf_and_display(file_path, file_description, column_names, nrows=10):
    """
    Reads an RRF file into a pandas DataFrame, displays the first N rows
    and column names, handling large files by reading only a specified number of rows.
    """
    print(f"Reading {file_description} from: {file_path}")
    try:
        df = pd.read_csv(
            file_path,
            sep='|',
            header=None,
            names=column_names, # Use the provided column names
            nrows=nrows,
            encoding='utf-8',
            on_bad_lines='skip'
        )

        print(f"Successfully read {file_description}.\n")
        print(f"First {nrows} rows of {file_description}:")
        display(df.head(nrows))

        print(f"\nColumn names (indices) of {file_description}:")
        print(df.columns.tolist())
        return df

    except FileNotFoundError:
        print(f"Error: The file '{file_path}' for {file_description} was not found.")
        return None
    except Exception as e:
        print(f"An error occurred while reading {file_description} from '{file_path}': {e}")
        return None
```

# Generate Summary Table

```python
import pandas as pd
import sys

summary_data = []

# Iterate through all variables in the current global scope
for var_name, var_obj in list(sys.modules[__name__].__dict__.items()):
    # Check if the variable is a pandas DataFrame and its name starts with 'df_'
    if var_name.startswith('df_') and isinstance(var_obj, pd.DataFrame):
        # Extract the base name (e.g., 'mrconso' from 'df_mrconso')
        base_name = var_name[len('df_'):]

        # Determine the original RRF file name
        # Convert to uppercase to match dictionary keys
        uppercased_base_name = base_name.upper()

        if uppercased_base_name + '.RRF' in rrf_column_definitions:
            df_name = uppercased_base_name + '.RRF'
        elif uppercased_base_name in rrf_column_definitions:
            df_name = uppercased_base_name
        else:
            # Handle potential MRSW_ to MRXW_ conversion if not found directly
            if uppercased_base_name.startswith('MRSW_'):
                corrected_name = uppercased_base_name.replace('MRSW_', 'MRXW_', 1)
                if corrected_name + '.RRF' in rrf_column_definitions:
                    df_name = corrected_name + '.RRF'
                elif corrected_name in rrf_column_definitions:
                    df_name = corrected_name
                else:
                    df_name = uppercased_base_name # Fallback if correction doesn't match
            else:
                df_name = uppercased_base_name # Fallback if no specific match

        num_rows = var_obj.shape[0]
        column_names = var_obj.columns.tolist()

        summary_data.append({
            'DataFrame Name': df_name,
            'Number of Rows': num_rows,
            'Column Names': column_names
        })

# Create the summary DataFrame
df_summary = pd.DataFrame(summary_data)

print("Summary of Loaded DataFrames:")
display(df_summary)
```
