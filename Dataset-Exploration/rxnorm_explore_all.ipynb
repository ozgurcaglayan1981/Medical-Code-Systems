# RxNorm Data Processing

```python
import os
import shutil
import zipfile

# Define paths
zip_file_source_dir = '/content/drive/MyDrive/05_data/RxNorm'
extract_to_dir = '/content/drive/MyDrive/05_data/RxNorm_extracted_data'

# Create the extraction directory if it doesn't exist
os.makedirs(extract_to_dir, exist_ok=True)

# Find the zip file in the source directory
zip_file_name = None
for f in os.listdir(zip_file_source_dir):
    if f.endswith('.zip'):
        zip_file_name = f
        break

if zip_file_name is None:
    print(f"Error: No zip file found in '{zip_file_source_dir}'")
else:
    zip_file_path = os.path.join(zip_file_source_dir, zip_file_name)
    print(f"Found zip file: '{zip_file_path}'")
    print(f"Extracting '{zip_file_name}' to '{extract_to_dir}'...")

    try:
        shutil.unpack_archive(zip_file_path, extract_to_dir)
        print("Extraction complete.")
    except Exception as e:
        print(f"Error during extraction: {e}")

# List all files in the extracted directory
rxnorm_files = []
if os.path.exists(extract_to_dir):
    rxnorm_files = [f for f in os.listdir(extract_to_dir) if os.path.isfile(os.path.join(extract_to_dir, f))]
    print(f"\nFiles in '{extract_to_dir}':")
    for f in rxnorm_files:
        print(f)
else:
    print(f"Extracted directory '{extract_to_dir}' not found.")

# Store the list of files for subsequent processing
# This variable will be used in the next steps to generate cells for each file.
# For now, we print it to confirm.
print(f"\nDiscovered RxNorm files: {rxnorm_files}")
```

This `RXNSAT.RRF` file from the `prescribe` subdirectory contains additional attributes for RxNorm concepts and atoms relevant to drug prescribing. These attributes might include specific strength values, units, or other details necessary for precise drug dispensing and administration, providing more granular information than the main `RXNSAT.RRF` for this specific use case.

## Column Descriptions

Similar to the main `RXNSAT.RRF` file, this file typically contains columns such as:

*   **`RXCUI`**: RxNorm Concept Unique Identifier.
*   **`LUI`**: Term Unique Identifier.
*   **`SUI`**: String Unique Identifier.
*   **`RXAUI`**: RxNorm Atom Unique Identifier.
*   **`STYPE`**: Source Type.
*   **`CODE`**: Source code of the attribute.
*   **`ATN`**: Attribute Name.
*   **`SAB`**: Source Abbreviation.
*   **`ATV`**: Attribute Value.
*   **`SUPPRESS`**: Suppression status.
*   **`CVF`**: Content View Flag.
