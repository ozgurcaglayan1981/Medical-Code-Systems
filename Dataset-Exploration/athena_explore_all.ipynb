# data.csv

## Overview

This `data.csv` file, identified from its content, primarily contains **HCPCS/CPT codes and their associated details**. It serves as a comprehensive listing for various medical procedures, services, and supplies, including modifiers, descriptions, and payment-related information. This file is crucial for understanding the billing and procedural aspects within a healthcare context.

## Column Descriptions

The `data.csv` file includes the following key columns, providing detailed information for each HCPCS/CPT entry:

*   **`HCPCS_Code`**: The unique identifier for a medical procedure, service, or supply.
*   **`Modifier`**: Codes that provide additional information about a service or procedure.
*   **`Description`**: A textual explanation of the HCPCS/CPT code.
*   **`Status_Code`**: Indicates the status or type of the code.
*   **`Is_Used_For_Medicare_Payment`**: Specifies if the code is used for Medicare payment.
*   **`Work_RVU`**: Work Relative Value Units, reflecting the physician's work effort.
*   **`NonFacility_PE_RVU`**: Non-Facility Practice Expense RVUs.
*   **`Is_NonFacility_NA_Indicator`**: Indicator for non-facility N/A.
*   **`Facility_PE_RVU`**: Facility Practice Expense RVUs.
*   **`Is_Facility_NA_Indicator`**: Indicator for facility N/A.
*   **`Malpractice_RVU`**: Malpractice expense RVUs.
*   **`Global_Period`**: Defines the post-operative period during which follow-up care is included in the surgical package.
*   **`Physician_Supervision`**: Indicates the level of physician supervision required.
*   **`Pre_Operative_Percent`**: Percentage of work RVUs attributed to pre-operative care.
*   **`Intra_Operative_Percent`**: Percentage of work RVUs attributed to intra-operative care.
*   **`Post_Operative_Percent`**: Percentage of work RVUs attributed to post-operative care.
*   **`Multiple_Procedure`**: Rules for multiple procedures.
*   **`Bilateral_Surgery`**: Indicator for bilateral surgery.
*   **`Assistant_at_Surgery`**: Rules for assistant at surgery.
*   **`Co_Surgeons_Modifier_62`**: Rules for co-surgeons (Modifier 62).
*   **`Team_Surgeons_Modifier_66`**: Rules for team surgeons (Modifier 66).
*   **`Used_for_APCs`**: Indicates if the code is used for Ambulatory Payment Classifications.
*   **`Endoscopy_Rule`**: Rules specific to endoscopy procedures.
*   **`NonFacility_PE_Used_For_OPPS_Payment_Amount`**: Non-facility practice expense used for OPPS payment.
*   **`Facility_PE_Used_For_OPPS_Payment_Amount`**: Facility practice expense used for OPPS payment.
*   **`Malpractice_Used_For_OPPS_Payment_Amount`**: Malpractice expense used for OPPS payment.

These columns collectively provide a detailed financial and operational context for each medical code.

```python
import pandas as pd
import os

athena_path = '/content/drive/MyDrive/05_data/ATHENA'
file_name = 'data.csv'
file_path = os.path.join(athena_path, file_name)

print(f"### Reading '{file_name}'\n")
try:
    df = pd.read_csv(file_path, nrows=100)
    print(f"Successfully loaded '{file_name}'. Displaying the first 10 rows:")
    print(df.head(10))
except Exception as e:
    print(f"Error reading '{file_name}': {e}")
```

# CONCEPT_ANCESTOR.csv

## Overview

The `CONCEPT_ANCESTOR.csv` file is fundamental for understanding the **hierarchical relationships between different concepts** within the vocabulary. It meticulously maps out parent-child (ancestor-descendant) connections, which are vital for semantic reasoning, aggregating data, and navigating complex medical terminologies. This file allows users to trace the lineage of concepts and understand their broader or narrower meanings.

## Column Descriptions

The `CONCEPT_ANCESTOR.csv` file contains the following critical columns to describe concept hierarchies:

*   **`ancestor_concept_id`**: The unique identifier of the higher-level (parent) concept in the hierarchy.
*   **`descendant_concept_id`**: The unique identifier of the lower-level (child) concept in the hierarchy.
*   **`min_levels_of_separation`**: The shortest distance (number of relationships) from the ancestor to the descendant.
*   **`max_levels_of_separation`**: The longest distance (number of relationships) from the ancestor to the descendant.

These columns enable comprehensive analysis of how concepts are related through various levels of abstraction.

```python
import pandas as pd
import os

athena_path = '/content/drive/MyDrive/05_data/ATHENA'
file_name = 'CONCEPT_ANCESTOR.csv'
file_path = os.path.join(athena_path, file_name)

print(f"### Reading '{file_name}'\n")
try:
    df = pd.read_csv(file_path, sep='\t', on_bad_lines='skip', engine='python', nrows=100)
    print(f"Successfully loaded '{file_name}'. Displaying the first 10 rows:")
    print(df.head(10))
except Exception as e:
    print(f"Error reading '{file_name}': {e}")
```

# DRUG_STRENGTH.csv

## Overview

The `DRUG_STRENGTH.csv` file is a critical component for detailed analysis of **drug dosages, formulations, and quantitative properties**. It provides precise information about the strength, form, and units of various drug concepts, making it indispensable for pharmacovigilance, dose calculations, and understanding the composition of medicinal products.

## Column Descriptions

The `DRUG_STRENGTH.csv` file includes comprehensive details about drug composition and strength:

*   **`drug_concept_id`**: The unique identifier for the drug concept itself.
*   **`ingredient_concept_id`**: The unique identifier for the active ingredient within the drug.
*   **`amount_value`**: The numerical value representing the amount of the ingredient.
*   **`amount_unit_concept_id`**: The unit of measure for the `amount_value` (e.g., mg, g).
*   **`numerator_value`**: The numerical value for the numerator in a ratio strength (e.g., in a concentration like 10mg/mL, this would be 10).
*   **`numerator_unit_concept_id`**: The unit of measure for the `numerator_value`.
*   **`denominator_value`**: The numerical value for the denominator in a ratio strength (e.g., in 10mg/mL, this would be 1).
*   **`denominator_unit_concept_id`**: The unit of measure for the `denominator_value`.
*   **`box_size`**: The number of units in a package or box.
*   **`valid_start_date`**: The date from which the drug strength information is valid.
*   **`valid_end_date`**: The date until which the drug strength information is valid.
*   **`invalid_reason`**: Indicates if the record is invalid and why (e.g., 'D' for deprecated).

# RELATIONSHIP.csv

## Overview

The `RELATIONSHIP.csv` file functions as a **lookup table for all defined relationship types** that can exist between concepts. It provides detailed descriptions and properties for each `relationship_id` found in the `CONCEPT_RELATIONSHIP.csv` file. This is crucial for fully understanding the nature and implications of semantic connections between concepts, including whether a relationship is hierarchical or has a defined ancestry.

## Column Descriptions

`RELATIONSHIP.csv` provides definitions for the types of relationships:

*   **`relationship_id`**: A unique short identifier for the relationship type (e.g., 'Is a', 'Maps to').
*   **`relationship_name`**: The full, human-readable name of the relationship.
*   **`is_hierarchical`**: A boolean (0 or 1) indicating if this relationship type denotes a hierarchical connection.
*   **`defines_ancestry`**: A boolean (0 or 1) indicating if this relationship type contributes to defining concept ancestry.
*   **`reverse_relationship_id`**: The `relationship_id` for the inverse of this relationship (e.g., if 'Has child' is the `relationship_id`, 'Has parent' would be its `reverse_relationship_id`).
*   **`relationship_concept_id`**: The unique identifier for the relationship type itself, treated as a concept.

This file is key to interpreting the `relationship_id` values used in other concept-related tables.

# CONCEPT.csv

## Overview

The `CONCEPT.csv` file is the **master list and core of the entire Athena vocabulary**. It contains the definition and properties of *every single concept* used across all vocabularies (e.g., SNOMED, RxNorm, ICD). Each row represents a unique concept, providing its name, domain, vocabulary, class, and status. This file is foundational for any analysis, mapping, or standardization effort within the OMOP Common Data Model, serving as the ultimate source of truth for all conceptual entities.

## Column Descriptions

The `CONCEPT.csv` file contains the most comprehensive details for every concept:

*   **`concept_id`**: The unique identifier for the concept.
*   **`concept_name`**: The primary, human-readable name or description of the concept.
*   **`domain_id`**: The high-level domain to which the concept belongs (e.g., 'Condition', 'Drug', 'Measurement').
*   **`vocabulary_id`**: The vocabulary from which the concept originates (e.g., 'SNOMED', 'RxNorm', 'ICD10CM').
*   **`concept_class_id`**: A more granular classification of the concept within its vocabulary (e.g., 'Clinical Drug', 'Race', 'Procedure').
*   **`standard_concept`**: An indicator ('S' for Standard, 'C' for Classification, 'N' for Non-standard) of whether the concept is a standard concept used for data storage.
*   **`concept_code`**: The original source code from the vocabulary (e.g., SNOMED CT code, RxNorm CUI).
*   **`valid_start_date`**: The date from which the concept is considered valid.
*   **`valid_end_date`**: The date until which the concept is considered valid.
*   **`invalid_reason`**: Indicates if the concept is invalid and why (e.g., 'D' for deprecated, 'U' for upgraded).

These columns provide a complete profile for each concept, enabling detailed semantic understanding and data mapping.

```python
import pandas as pd
import os

athena_path = '/content/drive/MyDrive/05_data/ATHENA'
file_name = 'CONCEPT.csv'
file_path = os.path.join(athena_path, file_name)

print(f"### Reading '{file_name}'\n")
try:
    df = pd.read_csv(file_path, sep='\t', on_bad_lines='skip', engine='python', nrows=100)
    print(f"Successfully loaded '{file_name}'. Displaying the first 10 rows:")
    print(df.head(10))
except Exception as e:
    print(f"Error reading '{file_name}': {e}")
```
